```python
from slot_util import load_slot_tables, print_df_attrs, get_reel_symbols, get_expanding_wild_window_df, get_expanding_wild_reelstop_df, get_line_pattern_df, get_expanding_wild_line_rtp_df
```

```python
reel_df, paytable_df, payline_df, _, game_settings = load_slot_tables("sheet/ExpandingWild.xlsx")
```

```python
print_df_attrs(reel_df)
display(reel_df)

print_df_attrs(paytable_df)
display(paytable_df)

print_df_attrs(payline_df)
display(payline_df)

print_df_attrs(game_settings)
display(game_settings)
```

```python
symbols = get_reel_symbols(reel_df)
symbols
```

```python
expanding_window_df = get_expanding_wild_window_df(reel_df, payline_df, game_settings["WindowHeight"])
```

```python
print_df_attrs(expanding_window_df)
display(expanding_window_df)
```

```python
reelstop_df = get_expanding_wild_reelstop_df(expanding_window_df, symbols)
```

```python
print_df_attrs(reelstop_df)
display(reelstop_df)
```

```python
line_df = get_line_pattern_df(reel_df, paytable_df, symbols)
expanding_line_df = get_expanding_wild_line_rtp_df(line_df, reelstop_df, paytable_df)
```

```python
print_df_attrs(expanding_line_df)
display(
    expanding_line_df.style.format({
        "RTP": "{:.2%}",
    })
)
```

```python
line_rtp = (
    expanding_line_df
    .groupby("Line", sort=False)["RTP"]
    .sum()
)

total_rtp = line_rtp.mean()
```

```python
print(f'RTP: {total_rtp:.2%}')
```
