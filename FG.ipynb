```python
from slot_util import load_slot_tables, print_df_attrs, get_reel_symbols, \
    get_reelstop_df, add_wild_combination_counts, add_scatter_window_counts, get_line_pattern_df, add_line_rtp, \
    get_fg_trigger_analysis
```

## Base Game

```python
reel_df, paytable_df, free_spins_df, game_settings = load_slot_tables("sheet/FG.xlsx")
```

```python
print_df_attrs(reel_df)
display(reel_df)

print_df_attrs(paytable_df)
display(paytable_df)

print_df_attrs(free_spins_df)
display(free_spins_df)

print_df_attrs(game_settings)
display(game_settings)
```

```python
symbols = get_reel_symbols(reel_df)
symbols
```

```python
reelstop_df = get_reelstop_df(reel_df, symbols)
```

```python
add_wild_combination_counts(reelstop_df, symbols)
add_scatter_window_counts(reelstop_df, window_height=game_settings["WindowHeight"])
```

```python
print_df_attrs(reelstop_df)
display(reelstop_df)
```

```python
line_df = get_line_pattern_df(reel_df, paytable_df, symbols, free_spins_df)
add_line_rtp(line_df, reelstop_df, paytable_df)
```

```python
print_df_attrs(line_df)
display(
    line_df.style.format({
        "RTP": "{:.2%}",
    })
)
```
